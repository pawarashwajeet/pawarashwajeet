## Vectorization, ufuncs and broadcasting
$$s  =\sum_{i=1}^{100} \sqrt{\frac{i\pi}{100}} \sin\left(\frac{i\pi}{100} \right)$$

```python
import math

x = 0; s = 0
for i in range(1,101):
  s = s + math.sqrt(x)*math.sin(x)
  x = x + 0.01*math.pi

  print(s)
```

The Vectorization version of Alogrithm

```python
import numpy as np

x = np.arange(0, 1.01*np.pi, 0.01*np.pi)
## arange function doesnt include last element in an array
print(np.sum(np.sqrt(x)*np.sin(x)))
```
