## NumPy: The Engine of Scientific Computing

In Fluids we deal with grids, fields and matrices.NumPy is the standard for manipulating these numerical arrays.

** Key Concept: Vectorization**,</br>
In Python, standard for loops are tool slow for HPC computing.
NumPy uses `Vectorization`, pushing goperations down to optimized C/C++/FORTRAN backened code. When working with million cell meshes, always vectorize your operations.
